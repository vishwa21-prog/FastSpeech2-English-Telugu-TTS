```python
!pip install -q datasets huggingface_hub
from datasets import load_dataset
from kaggle_secrets import UserSecretsClient
token = UserSecretsClient().get_secret("HF_TOKEN")

ds = load_dataset("ai4bharat/Svarah", split="test", token=token,
                  cache_dir="/kaggle/temp/hf_cache")
print(ds)

meta = ds.remove_columns(["audio_filepath"]).to_pandas()
meta["spk"] = (meta["gender"] + "|" + meta["age-group"] + "|" +
               meta["primary_language"] + "|" + meta["native_place_district"])

print("speakers (proxy):", meta["spk"].nunique())
print("total hours:", meta["duration"].sum() / 3600)

ok = meta[(meta.duration >= 1.5) & (meta.duration <= 15)]
print("clips kept:", len(ok), "| hours kept:", ok["duration"].sum() / 3600)
print(ok.groupby("spk")["duration"].sum().div(60).describe())  # minutes per speaker
```

```python
from pathlib import Path

roots = [
    Path("/kaggle/working"),
    Path("/kaggle/input"),
]

names = [
    "FastSpeech2_BACKUP.zip",
    "1000.pth.tar",
    "english_pretrained.wav",
    "telugu_pretrained.wav",
    "train.txt",
]

for root in roots:
    print(f"\nSearching {root}")

    for name in names:
        matches = list(root.rglob(name))

        print(f"\n{name}: {len(matches)} found")

        for p in matches[:10]:
            print(" ", p)
```
